## List Recap


Lists are collection types that store a group of values in an indexed manner. They are created using [] syntax and can be modified after creation.

```python
# Creating a list
list_example = [1, 2, 3, "John"]

# Common list methods
list_example.append(10)            # adds 10 to the end
list_example.pop()                 # removes and returns the last value
list_example.insert(1, "new")      # inserts "new" at index 1
list_example.remove("John")        # removes the first occurrence of "John"
list_example.extend([20, 30])      # adds multiple values at once
list_example.sort(reverse=False)   # sorts in ascending order (True for descending)
```

List slicing lets you extract a sub-list using list[start:end], where start is included and end is excluded.

```python
list_1 = ["1", "10", "3", "4", "5", "6", "7", "8", "9"]

list_1[2:6]   # from index 2 to 5 → ['3', '4', '5', '6']
list_1[:6]    # from start to index 5 → ['1', '10', '3', '4', '5', '6']
list_1[2:]    # from index 2 to the end → ['3', '4', '5', '6', '7', '8', '9']
```

````python
# Task 1
# Create a list of 5 vegetables, then:
vegetables = ['carrot', 'broccoli', 'spinach', 'potato', 'tomato']
# - Append 2 new vegetables and print the list
vegetables.append("cucumber")
vegetables.append("lettuce")
print(vegetables)
# - Remove the last vegetable using pop() and print the list
vegetables.pop()
print(vegetables)
# - Use slicing to print only the first 3 vegetables
print(vegetables[:3])
````

## Tuple Recap

Tuples are similar to lists, but they cannot be modified after creation (immutable). They are commonly used to return multiple values from a function in an uneditable format. Tuples are created using ().

```python
# creating a tuple
tuple_1 = (1, 2, 3)

# unpacking a tuple into variables (based on order)
x, y, z = tuple_1
# x = 1, y = 2, z = 3

# accessing elements (just like lists)
print(tuple_1[0])   # 1
```

You can't use methods like .append(), .pop(), or .remove() on tuples.

````python
# Task 2
# create a tuple containing your name, age, and city
my_info = ("Ranjan Singh", 21, "Kathmandu")
# unpack the tuple into 3 variables and print each one
name, age, city = my_info
print(name)
print(age)
print(city)
````

## If Statement

If Statements are similar to other languages in terms of functionality, it checks for a condition and executes the following statement if the condition is true.

The difference is in the syntax.

```cpp
// In c-syntax language:
int age = 23;

if (age<18){
  cout<<"You can't vote";
}else{
  cout<<"You can vote";
}
```

Python doesn't use curly braces to seperate code blocks. It instead uses identation(just spacing). As you can see in the example the if block uses some space. If you wrote if statement and print statment in the same indentation level it wouldn't work.

```python
# In python:
age = 23

if age<18:
  print("You can't vote") # print has space in the front, this indicates it is part of the if block instead of {} curly braces
else:
  print("You can vote")
```

```python
if age<18:
print("You can't vote") # This would cause an error due to lack of spacing or indenting.
```

## For Loop

For loops just like in other languages are available in python too. It is used to loop a specified number of times based on a given condition.

In other languages:
```cpp
for (int x; x<=10; x++){
  cout<<"The value of x"<<x;
}
```

In python:

```python
for x in range(10):
  print(f"The value of x is {x}")
```

Both of these loops print value from 0 to 9

Range can be further elaborated:

```python
range(0, 10, 2)
```
The first parameter determines the start of the value, the second parameter determines the last value, and the last parameter determines the step.

In this case the loop would start with value of x = 0, it would go until x<10, and it would increase by 2 every turn.

````python
# Task 10 Use for loop to print values from 0 to 15
for i in range(16):
    print(i)
````

## While Loop

While loop in python loops until a condition is met. You create a condition that eventually breaks and adjust the condition throughout the course of the loop.

```python

while a<10: # Loops until a<10
  print(a)
  a+=1 # a increases every turn
```

## More on For Loops

**Create a cell and try these out if you have already completed above tasks.**

We talked about expanding range before. This is infact very flexible:

```python
# range(start, stop, step) - with a custom step
for i in range(0, 10, 2):
    print(i)   # 0, 2, 4, 6, 8

# Counting backwards with a negative step
for i in range(10, 0, -1):
    print(i)   # 10, 9, 8, ... 1
```

While looping through a list with a for loop, if you also want index value, you can use enumerate():

```python
fruits = ["apple", "banana", "mango"]
for index, fruit in enumerate(fruits):
    print(index, fruit)
# 0 apple
# 1 banana
# 2 mango
```

You can loop through multiple lists together with zip:

```python
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]

for name, age in zip(names, ages):
    print(name, age)
```

You can use continue to skip an iteration of the loop:

```python
for i in range(5):
    if i == 2:
        continue    # skips the rest of this iteration
    print(i)
# Prints 0, 1, 3, 4
```

You can also use break to terminate the loop early:

```python
for i in range(10):
    if i == 5:
        break    # exits the loop completely
    print(i)
# Prints 0, 1, 2, 3, 4
```

Loops can be nested as well:

```python
# Useful for grids, matrices, or pairs
for i in range(3):
    for j in range(3):
        print(i, j)
```

## Missing Data

From using .info() method above, you can see that there are a lot of non-null values in some of the columns. That, means those columns have values that weren't entered in the data collection process. Real data comes with such kind of discrepancies you have to account for.

One of the easiest approach to deal with this problem is just to drop the dataset.

```python
df_clean = df.dropna() # Drops all rows with missing values
```

````python
# Task 4
# Count the missing values in each column using isnull().sum()
df.isnull().sum()

# Create a cleaned DataFrame by dropping rows with missing values:
cleaned_df = df.dropna()

# Print the shape of the DataFrame again to see how many rows you lost due to the drop command
cleaned_df.shape
cleaned_df.info()
````

## Selecting & Filtering

Pull specific column(s) by using square bracket. Multiple columns can be pulled by using another pair of square bracket similar to using a list.

```python
id_col = df["id"] # This will pull column named id

```

```python
passed_students = df[df["score"]>40] # This will filter out students with score more than 40
```

Multiple conditions can be combined as follows:

```python

```

Use & for AND logic and | for OR logic

````python
# Task 5
# Print just the "species" column
species_column = df['species']
print(species_column)

# Print the "species", "island", and "body_mass_g" columns together
multi_column = df[['species', 'island', 'body_mass_g']]
multi_column

table_describe = multi_column.describe()
table_describe
````

````python
# Print only the penguins heavier than 4500 grams
heavier_penguins = df[df['body_mass_g'] > 4500]
heavier_penguins
````

````python
# Print only the penguins from the island "Biscoe"
df[df['island'] == 'Biscoe']
````

````python
# Print penguins that are BOTH heavier than 4000g AND of species "Gentoo"
df[(df['body_mass_g'] > 4000) & (df['species'] == 'Gentoo')]
````

````python
# Print penguins from EITHER "Biscoe" OR "Dream" island
df[(df['island'] == 'Biscoe') | (df['island'] == 'Dream')]
````

## Group by

Just like GROUP BY in SQL, pandas allows you to group rows by certain category to obtain aggregated information.

```python
df.groupby("location")["salary"].mean() #Gives average salary per location
df.groupby("location").size() # How many students or person per location
```

````python
# Task 7
# Find the average body_mass_g for each species
df.groupby('species')['body_mass_g'].mean()
````

````python
# Find the average flipper_length_mm for each species
df.groupby('species')['flipper_length_mm'].mean()
````

````python
# Count how many penguins are on each island (groupby + size)
df.groupby('island').size()
````

````python
# Find the maximum body_mass_g for each species
df.groupby('species')['body_mass_g'].max()
````

## Combining it all

All of these concepts and features of df can be used to answer complex questions about the datset. You can use them to answer the questions below.

````python
# 1. Which species has the highest average body mass? use .idxmax() to return highest index value from grouped values.
df.groupby('species')['body_mass_g'].mean().idxmax()
````

````python
# 2. On which island do penguins have the longest average flipper length?
df.groupby('island')['flipper_length_mm'].mean().idxmax()
````

````python
# 3. Are Gentoo penguins heavier on average than Adelie penguins?

adelie_avg = df[df['species'] == 'Adelie']['body_mass_g'].mean()
gentoo_avg = df[df['species'] == 'Gentoo']['body_mass_g'].mean()

print("Adelie average:", adelie_avg)
print("Gentoo average:", gentoo_avg)
print("Gentoo heavier?", gentoo_avg > adelie_avg)
````

````python
# 4. How many penguins weigh more than the overall average body mass?
#    (hint: compute the mean first, then filter)
overall_avg = df['body_mass_g'].mean()
overall_avg
df[df['body_mass_g'] > overall_avg]
````

````python
# 5. For each species, what is the average mass of just the penguins on Biscoe island? 
# (hint: filter first, then groupby)
df[df['island'] == 'Biscoe'].groupby('species')['body_mass_g'].mean()
````
